```python
import pandas as pd
import re

# Step 1: Load your dataset


# Step 2: Clean year column headers
def clean_year_columns(col):
    if isinstance(col, str) and re.match(r'20\d{2}', col):
        return re.findall(r'20\d{2}', col)[0]  # Keep only the 4-digit year
    return col

df.columns = [clean_year_columns(col) for col in df.columns]

# Step 3: Confirm the result
print(df.columns.tolist())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# ─────────────────────────────────────────────
# Configuration
# ─────────────────────────────────────────────

la_codes = [
    "E07000146", "E07000147", "E07000143", "E07000144",
    "E07000149", "E07000148", "E07000145"
]

# ─────────────────────────────────────────────
# Melt to long format
# ─────────────────────────────────────────────
df_melted = df.melt(
    id_vars=["LA code", "LA name", "Sex", "Age"],
    var_name="Year",
    value_name="Population"
)
df_melted["Year"] = df_melted["Year"].astype(int)

# ─────────────────────────────────────────────
# Pre-aggregate for England-level dependency ratios
# ─────────────────────────────────────────────
def calculate_dependency_ratios(df_group):
    youth = df_group[df_group["Age"] <= 14]
    working = df_group[(df_group["Age"] >= 15) & (df_group["Age"] <= 64)]
    old = df_group[df_group["Age"] >= 65]

    youth_total = youth.groupby("Year")["Population"].sum()
    working_total = working.groupby("Year")["Population"].sum().replace(0, pd.NA).dropna()
    old_total = old.groupby("Year")["Population"].sum()

    youth_total = youth_total.reindex(working_total.index).fillna(0)
    old_total = old_total.reindex(working_total.index).fillna(0)

    oadr = (old_total / working_total) * 100
    ydr = (youth_total / working_total) * 100

    return pd.DataFrame({
        "Year": oadr.index,
        "Old-age Dependency Ratio (%)": oadr.values,
        "Youth Dependency Ratio (%)": ydr.values
    })

# England
england_df = calculate_dependency_ratios(df_melted)



# ─────────────────────────────────────────────
# Loop through selected LAs and generate plots
# ─────────────────────────────────────────────
for la_code in la_codes:
    la_df = df_melted[df_melted["LA code"] == la_code]
    la_name = la_df["LA name"].iloc[0]

    la_ratios = calculate_dependency_ratios(la_df)

    # Combine for range calculation
    all_values = pd.concat([
        la_ratios["Old-age Dependency Ratio (%)"],
        la_ratios["Youth Dependency Ratio (%)"],
        england_df["Old-age Dependency Ratio (%)"],
        england_df["Youth Dependency Ratio (%)"]
    ])
    y_min, y_max = all_values.min(), all_values.max()
    padding = (y_max - y_min) * 0.1 if y_max > y_min else 5
    y_lower = max(0, y_min - padding)
    y_upper = y_max + padding

    # Split estimates vs projections
    est_la = la_ratios[la_ratios["Year"] <= 2024]
    proj_la = la_ratios[la_ratios["Year"] >= 2024]
    est_eng = england_df[england_df["Year"] <= 2024]
    proj_eng = england_df[england_df["Year"] >= 2024]

    # ─────────────────────────────────────────────
    # Plotting
    # ─────────────────────────────────────────────
    fig, ax = plt.subplots(figsize=(16, 8))
    fig.patch.set_facecolor('white')

    # LA Lines (named)
    ax.plot(est_la["Year"], est_la["Old-age Dependency Ratio (%)"],
            label=f"{la_name} Old-age Dependency Ratio", color="black", linewidth=2.5)
    ax.plot(proj_la["Year"], proj_la["Old-age Dependency Ratio (%)"],
            color="black", linestyle="--", linewidth=2.5)
    ax.plot(est_la["Year"], est_la["Youth Dependency Ratio (%)"],
            label=f"{la_name} Youth-age Dependency Ratio", color="blue", linewidth=2.5)
    ax.plot(proj_la["Year"], proj_la["Youth Dependency Ratio (%)"],
            color="blue", linestyle="--", linewidth=2.5)

    # England Lines
    ax.plot(est_eng["Year"], est_eng["Old-age Dependency Ratio (%)"],
            label="England Old-age Dependency Ratio", color="darkgrey", linewidth=2.5)
    ax.plot(proj_eng["Year"], proj_eng["Old-age Dependency Ratio (%)"],
            color="darkgrey", linestyle="--", linewidth=2.5)
    ax.plot(est_eng["Year"], est_eng["Youth Dependency Ratio (%)"],
            label="England Youth-age Dependency Ratio", color="skyblue", linewidth=2.5)
    ax.plot(proj_eng["Year"], proj_eng["Youth Dependency Ratio (%)"],
             color="skyblue", linestyle="--", linewidth=2.5)

    ax.set_ylabel("Dependency Ratio (%)", fontsize=14)
    ax.set_xlabel("Year", fontsize=14)
    tick_positions = [year for year in la_ratios["Year"] if year % 2 == 0]
    ax.set_xticks(tick_positions)
    ax.set_xticklabels(tick_positions, fontsize=13)
    ax.axhline(0, color='grey', linestyle='--', linewidth=1)

    if 2024 in la_ratios["Year"].values:
        ax.axvline(x=2024, color='black', linestyle='--', linewidth=1.5)
        ax.axvspan(min(la_ratios["Year"]), 2024, color='lightgrey', alpha=0.2)
        ax.axvspan(2024, max(la_ratios["Year"]), color='lightblue', alpha=0.2)
        ax.text(2020, y_upper * 0.95, 'Estimates', fontsize=14)
        ax.text(2026, y_upper * 0.95, 'Projections', fontsize=14)

    ax.set_title(
        f"Youth and Old-age Dependency Ratios (per 100 Working-age): Estimates and Projections (2018–2043)\n{la_name} VS England",
        fontsize=18, pad=60, weight='bold'
    )
    plt.suptitle(
        "Dependency ratios calculated as the proportion of dependent age groups (0–14 and 65+) to the working-age population (15–64)",
        fontsize=14, y=0.91, x=0.5, ha='center'
    )

    ax.grid(True, linestyle='--', alpha=0.5)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.set_ylim(y_lower, y_upper)

    plt.legend(
        fontsize=13,
        loc='upper center',
        bbox_to_anchor=(0.5, -0.15),
        ncol=2,
        frameon=False
    )
    plt.tight_layout(rect=[0, 0, 1, 1.05])

    filename = f"Dependency_Ratios_1{la_name.replace(' ', '_').replace('/', '-')}.png"
    plt.savefig(os.path.join(plot_save_path, filename), dpi=300)
    plt.show()
```
